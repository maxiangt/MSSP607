```python
# 5. Comma Code practice project

def comma_code(item_list):

    if len(item_list) == 0:
        return ''
    elif len(item_list) == 1:
        return str(item_list[0])
    elif len(item_list) == 2:
        return str(item_list[0]) + ' and ' + str(item_list[1])
    else:
        result = ', '.join(item_list[:-1])
        result += ', and ' + item_list[-1]
        return result

spam = ['apples', 'bananas', 'tofu', 'cats']
print(comma_code(spam))
```
