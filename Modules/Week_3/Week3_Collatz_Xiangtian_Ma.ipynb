a. When does the code in a function execute: when the function is defined or called?
Only when a function is called does the code within the function body truly execute.


b. How many global scopes are there in a Python program? How many local scopes?
There is only one global scope, which is shared by the entire program.
There can be many local scopes - each time a function is called, a new local scope is created (even if the same function is called multiple times).

c. What happens to variables in a local scope when the function call returns?
After a function returns, its local scope is destroyed, and all the variables inside are cleared and no longer exist (that is, they can no longer be accessed from the outside).


d. How can you prevent a program from crashing when it gets an error?
Using try and except statements: Place the code that may cause errors in the try block. If an exception occurs, the program will jump to the except block to handle the error instead of crashing and exiting directly.

```python
def collatz(number):
    """Receive a number, process it based on parity and print/return the result"""
    if number % 2 == 0:
        # Even number: divided by 2
        result = number // 2
    else:
        # Odd numbers: Multiply by 3 and then add 1
        result = 3 * number + 1
    print(result)
    return result
```

```python
# Ask the user to enter the starting number and handle the error of non-integer input
try:
    num = int(input('Please enter an integer: '))
    while num != 1:
        num = collatz(num)
except ValueError:
    print('The input is invalid. Please enter an integer!')
```
