```python
import pandas as pd

df = pd.read_csv('sample_data/Student Performance Data.csv')
df.head()
```

```python
high_math = lambda data: data[data['math score'] >= 60]
result1 = high_math(df)
result1.head()
```

```python
add_avg = lambda data: data.assign(average_score=(data['math score'] + data['reading score'] + data['writing score']) / 3)
result2 = add_avg(df)
result2.head()
```

```python
by_gender = lambda data: data.groupby('gender')[['math score', 'reading score', 'writing score']].mean().reset_index()
result3 = by_gender(df)
result3
```
